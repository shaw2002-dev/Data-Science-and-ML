```python
"""
Measure of Dispersion -> How much spread the data is.

      - Range: Difference between max and min value in a dataset.

      - Variance: It is the average of the square difference between each data point and the mean.
       (sum(x-x_mean)^2)/N -> Population
       (sum(x-x_mean)^2)/n-1 -> Sample   [ n-1 is also called Bessel's coefficient.]

       => that n-1 is basically used in Sample coz during experiment they found that they are getting the nearest optimal solution reducing biasness.

       - Standard Deviation: Square root of Variance.

       - Coefficient of Variance: (SD/mean)*100

"""
```
