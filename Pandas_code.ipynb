```python
import pandas as pd
from sklearn.datasets import load_iris  # sklearn contains many dummy datasets.

iris = load_iris() # creating object here
print(iris.DESCR) # showing description of iris datasets.
```

```python
print(iris.data) # this shows us unstructured data.
```

```python
iris.target
```

```python
iris.feature_names
```

```python
train = pd.DataFrame(iris.data, columns = iris.feature_names) # passing rows and columns in a dataframe form. DataFrame helps dataset to be converted into structured data.
train
```

```python
train.head() # this will show u the first 5 data of a dataset.
```

```python
# how a column is being added in a dataset
train["Name_of_flower"] = iris.target
train.head(10) # this will show u the first 10 rows.
```

```python
train.tail() # this will show us the last 5 rows or data of a dataset.
```

```python
train.sample() # this will show us any one random rows from a dataset.
train.sample(3) # this will show us any three random rows from a dataset.
```

```python
train[["sepal length (cm)"]].head() # by this u can fetch the selected columns from the dataset that u have written inside that "".
# using just single square bracket can also show u the datatype that has been used in the dataset.
```

```python
train[train["Name_of_flower"] == 0] # this will show only those data for whose name of flower is 0.
```

```python
train[train["sepal length (cm)"] >7] # this will show u all the data containing sepal length only greater than 7.
```

```python
# give me rows with sepal length>7 and petal_length>7 from a dataset
train[(train["sepal length (cm)"] >7) & (train["petal length (cm)"] >7)] #using '&' not 'and' coz of boolean value.
```

```python
# '~' means it will give u the length which is less than 7 not greater than 7.
train [~(train["sepal length (cm)"] >7) | (train["petal length (cm)"] >7)]
```

```python
# setosa with petal width >2
train[(train["petal width (cm)"] >0.5) & (train["Name_of_flower"] == 0)]
```

```python
# GROUPBY - this will used for aggregation.
group = train.groupby("Name_of_flower").mean() # this will show u the mean of that particular column.
group
```

```python
# sum,max,min,count are some more aggregations of groupby.
# apply in pandas
train["sepal length (cm)"] = train["sepal length (cm)"].apply(lambda x: x*10) # it will modify the value of sepal length each time u run the code by multiplying with 10.
train.head()
```

```python
#CONCATENATION - merging two table into one.
employee1 = pd.DataFrame({
    'empid': [101,102,103,104,105],
    "Name": ["Sunita","Rohit","Dhoni","Kohli","Deepali"],
    "Department": ["HR","Finance","IT","Finance","IT"]
})

salary = pd.DataFrame({
    'empid': [106,107,108,109,110],
    "Salary": [500,650,2300,1250,345],
    "grade": ["A","B","C","D","E"]
})
concat_employee = pd.concat([employee1,salary],axis = 1)
concat_employee
```
