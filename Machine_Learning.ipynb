```python
"""
TYPES OF MACHINE LEARNING:

- Supervised Learning => It learns patterns from a labeled structured dataset.
- Semi Supervised Learning => Mixture of both supervised and unsupervised learning.
- Self supervised Learning => It learns patterns from unlabeled structured dataset.
- Unsupervised Learning => It learns patterns from independent features(column) of structured dataset.
- Reinforcement Learning => Learns from the hit and try or from losses.

"""
```

```python
"""
WHICH TYPES OF PROBLEM DOES MACHINE LEARNING SOLVE:

1) Regression based problems: when your target column is numerical then it is called Regression problem.
  Ex - House price prediction.
2) Classification Problems: when your target column is categorical then it is called Classification problem.
  Ex - Email spam detection.
2.1) Binary Classification Problem:
  If label column contains only 2 categories then that problem is Binary classification.
  Ex - Yes or No, True or False.
2.2) Multiclass Classification:
  If label column contains more than 2 categories then that problem is Multiclass classification.
  Ex - Iris dataset.

 Spam      Name_of_flower
  0           Setosa(0)
  1          Virginica(1)
  1           Setosa(0)
  0          Versicolor(2)
  1          Virginica(1)

2 classes    3 classes

2.3) Multilabel Classification:
  One record/row can have multiple labels(Problem with multiple label columns)(Neural Networks).
  Ex - Image Classification.

  apple banana cherry
    1     0      1
    0     1      0
    1     1      1

"""
```

```python
"""
DIFFERENT STAGES OF MACHINE LEARNING PIPELINE DEVELOPMENT:

1. Businees(problem) Understanding
2. Gathering data from real world
3. Data Pre-processing:
             Data cleaning(maintaining the authentication).
             Correlation check.
             Handling missing values.
             Handling Outliers.

4. Feature Engineering: Process of creating new features.
5. Encoding: Converting categorical columns to numerical column.
      Label Encoding and One-hot Encoding:
       |                         |
  setosa-(0)             setosa    virginica   versicolor
  virginica-(1)             1          0           0
  versicolor-(2)            0          1           0
                            0          0           1

6. Standardize the dataset.
7. Splitting of dataset into train and test data.
8. Model Selection followed by model training on train dataset.
9. Model Evaluation on test dataset.
10. Model Deployment.
11. Monitoring the model.
12. Retaining the model as accuracy become depleted.

House Prediction:
Input: Independent column
prediction: label column

"""
```
